```python
from xarm.wrapper import XArmAPI
import time

# Replace with your xArm's IP address
arm = XArmAPI('192.168.1.227')

# Enable motion and clear any existing errors
arm.motion_enable(True)
arm.clean_error()

# Set mode and state for motion
arm.set_mode(0)
arm.set_state(0)
time.sleep(1)

# Reset the linear track to origin
code = arm.set_linear_track_back_origin(wait=True)
print(f'Set linear track back to origin, code={code}')

# Verify if the linear track is at zero position
code, status = arm.get_linear_track_on_zero()
print(f'Linear track on zero point, code={code}, status={status}')
```

```python
pos = arm.get_linear_track_pos()
print('Current linear track position:', pos)
```

```python
# 1. Enable the linear track
arm.set_linear_track_enable(True)

# 2. Set a safe speed
arm.set_linear_track_speed(100)  # mm/s

# 3. Manually jog to 0 mm
code = arm.set_linear_track_pos(0, wait=True)
print(f"Move to 0 mm code: {code}")

# 4. Check position
code, pos = arm.get_linear_track_pos()
print(f"After move, linear track position: {pos}")
```
