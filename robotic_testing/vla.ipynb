# VLA control

Free-form control of the arm from a text instruction (ANDY's section).

The planner never touches the arm directly. It emits a list of actions from a fixed
vocabulary, and `VLARobot` (in `common/robot_test.py`) validates the whole plan before
anything moves:

- only actions in `VLARobot.ACTION_SPECS` exist, with typed and bounded parameters
- every translation is capped at the single-step limit, so a long move has to be
  expressed as several short ones
- the resulting TCP path is simulated against the safety envelope in the arm config
  (`safety_dict`), and the plan is rejected in full if any step leaves it
- `xArm.move_relative` re-checks the live pose on every single move, so the envelope
  holds even if the plan simulation could not predict a pose

The envelope is deliberately conservative: it is the bounding volume of the poses the
platform is already known to reach safely. Widen it in the arm config, and only after
physically checking the volume you are opening up is clear.

## Action schema for the planner

The vocabulary lives in `VLARobot.ACTION_SPECS`. This flat pydantic model mirrors it for
structured output; the assertions below fail loudly if the two ever drift apart, so a new
action added in `robot_test.py` cannot be silently unavailable here.


```python
from typing import List, Literal, Optional
from pydantic import BaseModel, Field


class RobotAction(BaseModel):
    """One robot action. Only the parameters belonging to the chosen action are filled in."""
    action: Literal[
        "get_state",
        "move_relative",
        "rotate_relative",
        "move_to_named_position",
        "home",
        "open_gripper",
        "close_gripper",
        "set_gripper",
        "pick_object",
        "place_object",
        "wave",
        "wait",
    ]
    dx: Optional[float] = Field(default=None, description="move_relative: offset along x in mm")
    dy: Optional[float] = Field(default=None, description="move_relative: offset along y in mm")
    dz: Optional[float] = Field(default=None, description="move_relative: offset along z in mm, positive is up")
    droll: Optional[float] = Field(default=None, description="rotate_relative: roll offset in degrees")
    dpitch: Optional[float] = Field(default=None, description="rotate_relative: pitch offset in degrees")
    dyaw: Optional[float] = Field(default=None, description="rotate_relative: yaw offset in degrees")
    name: Optional[str] = Field(default=None, description="move_to_named_position: name of the position")
    opening_mm: Optional[float] = Field(default=None, description="set_gripper: gripper opening in mm")
    descend_mm: Optional[float] = Field(default=None, description="pick_object/place_object: vertical approach in mm")
    grip_opening_mm: Optional[float] = Field(default=None, description="pick_object: opening to close to, in mm")
    times: Optional[int] = Field(default=None, description="wave: number of cycles")
    amplitude_deg: Optional[float] = Field(default=None, description="wave: swing amplitude in degrees")
    seconds: Optional[float] = Field(default=None, description="wait: how long to wait")


class RobotPlan(BaseModel):
    reasoning: str = Field(description="Brief explanation of the intended robot procedure.")
    actions: List[RobotAction]


# keep this schema and the executable vocabulary in lockstep
_schema_actions = set(RobotAction.model_fields['action'].annotation.__args__)
_registry_actions = set(VLARobot.ACTION_SPECS)
assert _schema_actions == _registry_actions, \
    f'schema/registry mismatch: {_schema_actions ^ _registry_actions}'

_schema_params = set(RobotAction.model_fields) - {'action'}
_registry_params = {p for spec in VLARobot.ACTION_SPECS.values() for p in spec['params']}
assert _schema_params == _registry_params, \
    f'parameter mismatch: {_schema_params ^ _registry_params}'

print(f'{len(_registry_actions)} actions, {len(_registry_params)} parameters, in sync')
```

## Query the planner

The system prompt is generated from the action registry plus the arm's *live* limits and
current pose, so the planner sees the real numbers it has to plan within instead of
guessing.

`image_path` is optional and unused for now: instructions come in as text. When the
camera is pointed at the workspace, pass a frame and the planner can reason about what it
sees alongside the state read off the arm.


```python
print(vla.system_prompt())
```

```python
def plan_robot_actions(query: str, image_path: str = None, extra_rules: str = "") -> RobotPlan:
    """Turn a text instruction into a validated-by-schema robot plan."""
    contents = [f"""{vla.system_prompt(extra_rules=extra_rules)}

USER REQUEST:
{query}
"""]

    if image_path:
        with open(image_path, 'rb') as image_file:
            contents.append(types.Part.from_bytes(
                data=image_file.read(),
                mime_type='image/jpeg',
            ))

    response = client.models.generate_content(
        model=MODEL,
        contents=contents,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=RobotPlan,
            temperature=0,
        ),
    )
    return RobotPlan.model_validate_json(response.text)


def run_query(query: str, dry_run=True, confirm=False, image_path=None, extra_rules=""):
    """
    Plan a request and hand it to the validating executor.

    Defaults to a dry run: look at the plan, then re-run with dry_run=False.
    """
    plan = plan_robot_actions(query, image_path=image_path, extra_rules=extra_rules)
    print('Reasoning:', plan.reasoning, end='\n\n')

    if not plan.actions:
        print('The planner returned no actions -- see the reasoning above.')
        return plan, []

    actions = [action.model_dump(exclude_none=True) for action in plan.actions]
    results = vla.execute_plan(actions, dry_run=dry_run, confirm=confirm)
    return plan, results
```
