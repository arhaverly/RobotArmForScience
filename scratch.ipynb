```python
import subprocess
import re, os
from utils.utils import adb_pull_latest_file, get_project_path
def adb_pull_latest_file(adb_path, android_dir, android_phone_id, local_dir):
    # Ensure the Android directory path ends with a '/'
    if not android_dir.endswith('/'):
        android_dir += '/'

    # Attempt to get a list of all files in the Android directory, sorted by last modification time
    try:
        cmd = [adb_path, '-s', android_phone_id, "shell", 'ls', '-lt', f'\"{android_dir}\"']
        adb_ls_output = subprocess.check_output(cmd, stderr=subprocess.STDOUT).decode("utf-8").strip()
    except subprocess.CalledProcessError as e:
        raise RuntimeError(f"Failed to execute ADB command to list files: {e.output.decode('utf-8')}")
    # Parse the output of the 'ls' command to get file names
    android_files = adb_ls_output.split("\n")

    # Filter the list to only include files that match a specific pattern (in this case, JPEG files with a certain naming convention)
    android_files = [f.split()[-1] for f in android_files if re.match(r'\d{8}_\d{6}_\w+\.jpg', f.split()[-1])]

    # Raise an error if there are no files to pull
    if not android_files:
        raise ValueError("No matching files found to pull.")

    # Get the name of the latest file based on the sorted 'ls' output
    latest_file = android_files[0]

    # Get a list of files already existing in the local directory
    local_files = os.listdir(local_dir)

    # Pull the latest file only if it doesn't already exist in the local directory
    if latest_file not in local_files:
        try:
            subprocess.run([adb_path, "-s", android_phone_id, "pull", f"{android_dir}{latest_file}", f"{local_dir}/{latest_file}"], check=True)
        except subprocess.CalledProcessError as e:
            raise RuntimeError(f"Failed to execute ADB pull command: {e}")
        return f"{local_dir}/{latest_file}"

ADB_PATH = "C:/Program Files/platform-tools/adb.exe"
ANDROID_PHONE_DIR = 'sdcard/Download/Meta View'
LOCAL_DIR = f'{get_project_path()}/gpt4-image-api/images'  # Assuming get_project_path() is defined
ANDROID_PHONE_ID = 'R92W60YQZ3M'

# Test the function
try:
    adb_pull_latest_file(ADB_PATH, ANDROID_PHONE_DIR, ANDROID_PHONE_ID, LOCAL_DIR)
except subprocess.CalledProcessError as e:
    raise RuntimeError(f"Failed to execute ADB command to list files: {e.output.decode('utf-8')}")

```
